```python
import math

# Board cells: empty = ' ', player 'x', opponent 'o'
board = [' ' for _ in range(9)]

def print_board():
  for i in range(3):
    # Fix: Corrected indexing for the board display
    print(board[3 * i], '|', board[3 * i + 1], '|', board[3 * i + 2])
    if i < 2:
      print('----------')

def is_winner(b, player):
  # Winning combinations in Tic-Tac-Toe
  wins = [
      (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
      (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
      (0, 4, 8), (2, 4, 6)
  ]
  for a, b_, c in wins:
    # Fix: Added b[c] to the winning condition and used correct boolean 'True'
    if b[a] == b[b_] == b[c] == player:
      return True
  return False  # Fix: Used correct boolean 'False'

def is_board_full(b):
  # Fix: Renamed function to match common usage and call site
  return ' ' not in b

def minimax(b, depth, is_maximizing):
  # Fix: Added colon, renamed parameter for clarity/consistency with internal logic
  """
  minimax recursive function:
  - is_maximizing: True if the current turn is maximizer (x), else minimizer (o)
  - Returns the best score from the perspective of the maximizer.
  """
  if is_winner(b, 'x'):
    return 10 - depth  # maximizer wins
  if is_winner(b, 'o'):
    return depth - 10  # minimizer wins
  if is_board_full(b):
    return 0  # draw

  if is_maximizing:
    best_score = -math.inf
    for i in range(9):
      if b[i] == ' ':
        b[i] = 'x'  # Try this move
        # Fix: depth + 1 and False for minimizer's turn
        score = minimax(b, depth + 1, False)
        b[i] = ' '  # Undo move
        # Fix: Corrected variable name case
        best_score = max(score, best_score)
    return best_score  # Fix: Corrected 'Return' to 'return' and indentation
  else:  # Fix: Corrected 'Else:' to 'else:'
    best_score = math.inf
    for i in range(9):
      if b[i] == ' ':
        b[i] = 'o'  # Try this move (opponent is 'o')
        score = minimax(b, depth + 1, True)  # Recursive for maximizer
        b[i] = ' '  # Undo move
        best_score = min(score, best_score)
    return best_score  # Fix: Corrected indentation

def best_move():
  """
  find the best move for maximizer (X).
  """
  best_score = -math.inf
  move = -1
  for i in range(9):
    if board[i] == ' ':  # Fix: Corrected 'bored' to 'board'
      board[i] = 'x'  # Fix: Consistent player symbol 'x'
      # Fix: Call minimax with False for the *next* player (minimizer)
      score = minimax(board, 0, False)
      board[i] = ' '  # Fix: Undo move correctly
      if score > best_score:
        best_score = score
        move = i
  return move

def play_game():
  """
  simple game loop: Human (O) vs AI (X using minimax).
  """
  print("Board positions are numbered 0 to 8:")
  print("0 | 1 | 2 ")
  print("----------")
  print("3 | 4 | 5")
  print("----------")
  print("6 | 7 | 8")
  print()
  while True:
    # AI move
    move = best_move()
    # The best_move function should always find a move if the board isn't full or a winner isn't determined yet
    # If it returns -1, it implies no valid moves or a tie condition, which is also handled by is_board_full.
    # This condition can be refined based on minimax return values if needed.
    if move == -1 and is_board_full(board):
        print("Game over! It's a tie.")
        break
    elif move == -1: # In case best_move returns -1 unexpectedly before board is full
        print("AI could not find a move. Game error.")
        break

    board[move] = 'x'  # Fix: Consistent player symbol 'x'
    print("AI (x) plays move:", move)
    print_board()  # Fix: Corrected function name

    if is_winner(board, 'x'):  # Fix: Consistent player symbol 'x'
      print("AI (x) wins!")  # Fix: Consistent player symbol 'x'
      break
    if is_board_full(board):  # Fix: Corrected function name
      print("Game over! It's a tie.")
      break

    # Human move
    while True:
      try:
        human_move = int(input("Enter your move (0-8):"))
        # Fix: Check for ' ' for empty space
        if 0 <= human_move <= 8 and board[human_move] == ' ':
          board[human_move] = 'o'  # Fix: Consistent player symbol 'o'
          break  # Break from inner while loop (human input)
        else:
          print("Invalid move. Try again.")
      except ValueError:  # Fix: Corrected 'excpt' to 'except'
        print("Please enter a valid number.")

    # These checks should be at the main game loop level after a human move
    print_board()
    if is_winner(board, 'o'):  # Fix: Added colon and consistent player symbol 'o'
      print("You win!")
      break
    if is_board_full(board):  # Fix: Corrected function name
      print("Game over! It's a Tie.")
      break

# Fix: Corrected indentation for the main execution block
if __name__ == "__main__":
  play_game()
```
